```python
import torch
from torch import nn
from torch import optim
import numpy as np
import PIL
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader
```

```python
dataset = "D:\\wahab stuff\\wahab code\\Draw-C-Ai\\data\\raw"
```

```python
class CNN(nn.Module):
    def __init__(self):
        super().__init__() 
        self.conv2d1 = nn.Conv2d(in_channels = 1, out_channels = 16,kernel_size = 3)
        self.relu = nn.ReLU()
        self.pool = nn.MaxPool2d(kernel_size = 2 , stride = 2)
        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(in_features = 2704 ,out_features = 10)

    def forward(self,x):
        x = self.conv2d1(x)
        x = self.relu(x)
        x = self.pool(x)
        x = self.flatten(x)
        x = self.fc1(x)
        return x
```

```python
import os 
import json
import glob
import numpy as np

dataset_dir = r"D:\wahab stuff\wahab code\Draw-C-Ai\dataset_v1\raw"
label_map = {
    "smile": 0, "square": 1, "triangle": 2, "heart": 3,
    "fire": 4, "thumbs_up": 5, "zap": 6, "x": 7
}

x_list, y_list = [], []
for filepath in glob.glob(os.path.join(dataset_dir, "*.json")):
    with open(filepath, "r") as f:
        data = json.load(f)
        x_list.append(data["vector"])
        y_list.append(label_map[data["label"]])

x_raw = np.array(x_list, dtype=np.float32)
y_raw = np.array(y_list, dtype=np.int64)
```

```python
x_tensor = torch.tensor(x_raw, dtype=torch.float32)
x_tensor = x_tensor.view(-1, 1, 32, 32)
y_tensor = torch.tensor(y_raw, dtype=torch.long)
```

```python
model = CNN()
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(),lr = 0.001)
output = model(x_tensor)
loss = criterion(output,y_tensor)
```

```python
epochs = 100
for epoch in range(epochs):
    optimizer.zero_grad()
    outputs = model(x_tensor)
    loss = criterion(outputs, y_tensor)
    loss.backward()
    optimizer.step()
    
    if (epoch + 1) % 10 == 0:
        print(f"Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}")
```

```python
torch.save(model.state_dict(),"weights/cnn_model.pth")
```
