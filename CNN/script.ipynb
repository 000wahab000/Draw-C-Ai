```python
class CNN(nn.Module):
    def __init__(self):
        super().__init__() 
        self.conv2d1 = nn.Conv2d(in_channels = 1, out_channels = 16,kernel_size = 3)
        self.relu = nn.ReLU()
        self.pool = nn.MaxPool2d(kernel_size = 2 , stride = 2)
        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(in_features = 3600 ,out_features = 8)

    def forward(self,x):
        x = self.conv2d1(x)
        x = self.relu(x)
        x = self.pool(x)
        x = self.flatten(x)
        x = self.fc1(x)
        return x
```
